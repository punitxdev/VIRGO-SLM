```python
import re

CATEGORY_PATTERNS = {
    category: re.compile(
        r"\b(?:"
        + "|".join(
            re.escape(keyword.lower())
            for keyword in keywords
            if keyword.strip()
        )
        + r")\b",
        re.IGNORECASE
    )
    for category, keywords in CATEGORY_KEYWORDS.items()
}


def classify_text(text, threshold_ratio=0.005):
    if not isinstance(text, str) or not text.strip():
        return "general"

    word_count = len(text.split())

    if word_count == 0:
        return "general"

    scores = {
        category: len(pattern.findall(text))
        for category, pattern in CATEGORY_PATTERNS.items()
    }

    best_cat = max(scores, key=scores.get)
    best_score = scores[best_cat]

    threshold = max(
        1,
        int(word_count * threshold_ratio)
    )

    if best_score < threshold:
        return "general"

    return best_cat
```

```python
import re
import unicodedata


def preprocess_text(text):
    if not isinstance(text, str):
        return None

    text = text.strip()

    if not text:
        return None

    # Unicode normalization
    text = unicodedata.normalize("NFKC", text)

    # Normalize line endings
    text = text.replace("\r\n", "\n")
    text = text.replace("\r", "\n")

    # Remove invisible/control characters
    text = "".join(
        char for char in text
        if char == "\n"
        or char == "\t"
        or unicodedata.category(char)[0] != "C"
    )

    # Preserve tabs
    text = text.replace("\t", " <tab> ")

    # Remove excessive spaces
    text = re.sub(r" {2,}", " ", text)

    # Clean spaces around newlines
    text = re.sub(r" *\n *", "\n", text)

    # Limit excessive newlines
    text = re.sub(r"\n{3,}", "\n\n", text)

    # Limit punctuation spam
    text = re.sub(r"!{5,}", "!!!!", text)
    text = re.sub(r"\?{5,}", "????", text)
    text = re.sub(r"\.{6,}", ".....", text)

    # Remove very short documents
    if len(text.split()) < 50:
        return None

    # Preserve newlines
    text = text.replace("\n", " <newline> ")

    # Document boundaries
    text = f"<bos> {text.strip()} <eos>"

    return text
```

```python
import pprint

# 2 Billion Token Budget
TOKEN_BUDGET = 2_000_000_000 

CATEGORY_RATIO = {
    "general": 0.40,
    "physics": 0.08,
    "mathematics": 0.08,
    "computer_science": 0.10,
    "biology": 0.06,
    "chemistry": 0.05,
    "engineering": 0.06,
    "history": 0.05,
    "economics": 0.04,
    "social_science": 0.03,
    "philosophy": 0.02,
    "literature": 0.03,
}

TOKEN_LIMITS = {}
for category, limit_ratio in CATEGORY_RATIO.items():
    TOKEN_LIMITS[category] = int(limit_ratio*TOKEN_BUDGET)

pprint.pp(TOKEN_LIMITS)
```

```python

WRITE_BUFFER_SIZE = 1_000_000
PROGRESS_EVERY = 1_000
CHECKPOINT_EVERY = 10_000

CHECKPOINT_PATH = "dataset_checkpoint.json"
```

```python
# TOKEN BUFFERS
# ============================================================

token_buffers = {
    category: []
    for category in TOKEN_LIMITS
}
```

```python
# FLUSH BUFFER
# ============================================================

def flush_buffer(category):

    if not token_buffers[category]:
        return

    tokens = np.asarray(
        token_buffers[category],
        dtype=np.uint16
    )

    tokens.tofile(bin_files[category])

    token_buffers[category].clear()
```

```python
# ============================================================
# SAVE CHECKPOINT
# ============================================================

def save_checkpoint():

    checkpoint = {
        "rows_seen": rows_seen,
        "docs_processed": docs_processed,
        "token_counts": token_counts,
    }

    with open(
        CHECKPOINT_PATH,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            checkpoint,
            f,
            indent=4
        )

    print(
        f"\n💾 Checkpoint saved | "
        f"Rows: {rows_seen:,} | "
        f"Docs: {docs_processed:,}"
    )
```

```python
# CATEGORY STATUS
# ============================================================

def print_category_status():

    print("\n")
    print("=" * 90)
    print("🌌 VIRGO DATASET STATUS")
    print("=" * 90)

    for category in TOKEN_LIMITS:

        processed = token_counts[category]

        limit = TOKEN_LIMITS[category]

        remaining = max(
            0,
            limit - processed
        )

        percentage = (
            processed / limit
        ) * 100

        print(
            f"{category:20s} | "
            f"{processed / 1e6:8.2f}M / "
            f"{limit / 1e6:8.2f}M | "
            f"Remaining: {remaining / 1e6:8.2f}M | "
            f"{percentage:6.2f}%"
        )

    total_processed = sum(
        token_counts.values()
    )

    total_remaining = (
        TOKEN_BUDGET
        - total_processed
    )

    print("-" * 90)

    print(
        f"TOTAL                | "
        f"{total_processed / 1e9:.3f}B / "
        f"{TOKEN_BUDGET / 1e9:.3f}B | "
        f"Remaining: {total_remaining / 1e9:.3f}B"
    )

    print("=" * 90)
```

```python
import os
import json
import numpy as np
from tqdm.auto import tqdm


# ============================================================
# LOAD CHECKPOINT
# ============================================================

if os.path.exists(CHECKPOINT_PATH):

    print("🔄 Loading checkpoint...")

    with open(
        CHECKPOINT_PATH,
        "r",
        encoding="utf-8"
    ) as f:

        checkpoint = json.load(f)

    rows_seen = checkpoint["rows_seen"]
    docs_processed = checkpoint["docs_processed"]
    token_counts = checkpoint["token_counts"]

    print(
        f"✅ Resume from row: {rows_seen:,}"
    )

    dataset_stream = dataset.skip(rows_seen)

else:

    print("🚀 Starting new dataset build")

    rows_seen = 0
    docs_processed = 0
    dataset_stream = dataset


# ============================================================
# ACTIVE CATEGORIES
# ============================================================

active_categories = {
    cat
    for cat in TOKEN_LIMITS
    if token_counts[cat] < TOKEN_LIMITS[cat]
}

print(
    f"🎯 Active categories: "
    f"{len(active_categories)}/{len(TOKEN_LIMITS)}"
)

print(active_categories)


# ============================================================
# PROGRESS BAR
# ============================================================

total_processed = sum(
    token_counts.values()
)

pbar = tqdm(
    total=TOKEN_BUDGET,
    initial=total_processed,
    desc="🌌 Virgo Dataset"
)


# ============================================================
# MAIN DATASET LOOP
# ============================================================

try:

    for row in dataset_stream:

        rows_seen += 1

        raw_text = row["text"]

        category = classify_text(
            raw_text,
            0.015
        )


        # ====================================================
        # SKIP COMPLETED CATEGORY
        # ====================================================

        if category not in active_categories:
            continue


        text = preprocess_text(raw_text)

        if text is None:
            continue


        tokens = tokenizer.encode(text).ids


        remaining = (
            TOKEN_LIMITS[category]
            - token_counts[category]
        )

        tokens = tokens[:remaining]


        token_buffers[category].extend(tokens)

        token_counts[category] += len(tokens)

        pbar.update(len(tokens))

        docs_processed += 1


        # ====================================================
        # CATEGORY COMPLETED
        # ====================================================

        if (
            token_counts[category]
            >= TOKEN_LIMITS[category]
        ):

            flush_buffer(category)

            active_categories.remove(category)

            print(
                f"\n✅ {category.upper()} COMPLETED"
            )

            print(
                f"📦 Tokens: "
                f"{token_counts[category] / 1e6:.2f}M"
            )

            print(
                f"🎯 Remaining categories: "
                f"{len(active_categories)}"
            )


        # ====================================================
        # FLUSH BUFFER
        # ====================================================

        elif (
            len(token_buffers[category])
            >= WRITE_BUFFER_SIZE
        ):

            flush_buffer(category)


        # ====================================================
        # LIVE PROGRESS
        # ====================================================

        if (
            docs_processed
            % PROGRESS_EVERY
            == 0
        ):

            total_tokens = sum(
                token_counts.values()
            )

            remaining_tokens = (
                TOKEN_BUDGET
                - total_tokens
            )

            pbar.set_postfix({

                "docs":
                    f"{docs_processed:,}",

                "total":
                    f"{total_tokens / 1e9:.3f}B",

                "remaining":
                    f"{remaining_tokens / 1e9:.3f}B",

                "active":
                    len(active_categories),

                "current":
                    category,

                "cat":
                    f"{token_counts[category] / 1e6:.1f}M",

                "cat_left":
                    f"{(
                        TOKEN_LIMITS[category]
                        - token_counts[category]
                    ) / 1e6:.1f}M"
            })


        # ====================================================
        # CHECKPOINT
        # ====================================================

        if (
            docs_processed
            % CHECKPOINT_EVERY
            == 0
        ):

            # Flush ONLY active categories
            for cat in active_categories:

                flush_buffer(cat)

            save_checkpoint()

            print_category_status()


        # ====================================================
        # FINISHED
        # ====================================================

        if not active_categories:

            print(
                "\n🎉 ALL CATEGORY LIMITS COMPLETED"
            )

            break


# ============================================================
# SAFE MANUAL STOP
# ============================================================

except KeyboardInterrupt:

    print(
        "\n⚠️ Manual stop detected"
    )

    for cat in active_categories:
        flush_buffer(cat)

    save_checkpoint()

    print("💾 Progress safely saved")
```

```python
# ============================================================
# FINAL SAVE
# ============================================================

finally:

    print(
        "\n💾 Flushing token buffers..."
    )

    for category in TOKEN_LIMITS:

        flush_buffer(
            category
        )

    save_checkpoint()

    print_category_status()

    for file in bin_files.values():

        file.close()

    pbar.close()

    print(
        "\n✅ Virgo dataset state saved safely"
    )
```

```python
import numpy as np
import random

BIN_PATH = "/kaggle/working/token_bins/mathematics.bin"

data = np.memmap(
    BIN_PATH,
    dtype=np.uint16,
    mode="r"
)

bos_id = tokenizer.token_to_id("<bos>")
eos_id = tokenizer.token_to_id("<eos>")

# ============================================================
# FIND DOCUMENT BOUNDARIES
# ============================================================

bos_positions = np.where(data == bos_id)[0]
eos_positions = np.where(data == eos_id)[0]

print(f"📄 Documents found: {len(bos_positions):,}")

# ============================================================
# RANDOM DOCUMENT
# ============================================================

while True:

    bos_pos = random.choice(bos_positions)

    eos_idx = np.searchsorted(
        eos_positions,
        bos_pos
    )

    if eos_idx < len(eos_positions):

        eos_pos = eos_positions[eos_idx]

        if eos_pos > bos_pos:
            break

# ============================================================
# EXTRACT ONE DOCUMENT
# ============================================================

sample_ids = data[
    bos_pos + 1:eos_pos
]

text = tokenizer.decode(
    sample_ids.tolist(),
    skip_special_tokens=False
)

# Restore formatting
text = text.replace("<newline>", "\n")
text = text.replace("<tab>", "\t")

print(f"🎲 Document start token: {bos_pos:,}")
print(f"📏 Document tokens: {len(sample_ids):,}")
print("=" * 100)
print(text)
```
