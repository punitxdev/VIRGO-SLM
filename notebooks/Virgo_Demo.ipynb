# ♍ Virgo SLM V1.0 - Interactive Demonstration

Welcome to the official demonstration notebook for **Virgo SLM V1.0**, a 120M parameter Small Language Model built completely from scratch by Punit Kumar Kashyap (Indian Institute of Technology Dharwad).

In this notebook, we will:
1. Initialize the Virgo 120M Transformer architecture.
2. Load pre-trained/instruct checkpoint weights.
3. Perform interactive token generation with customizable sampling parameters.

```python
# Step 1: Import Dependencies & Set Up System Path
import sys
import os
import torch
import torch.nn.functional as F

sys.path.append('..')
from webapp.model import VirgoModel

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Using compute device: {device}')
```

```python
# Step 2: Initialize Model Architecture (120M Parameters)
model = VirgoModel(
    vocab_size=45000,
    d_model=768,
    num_heads=12,
    num_layers=12,
    d_ff=3072,
    max_seq_length=1024,
    dropout=0.0
).to(device)

total_params = sum(p.numel() for p in model.parameters())
print(f'Virgo Model Architecture Initialized! Total parameters: {total_params:,}')
```

```python
# Step 3: Load Checkpoint Weights
ckpt_path = '../trained_models/virgo_instruct_v2.pt'
if os.path.exists(ckpt_path):
    checkpoint = torch.load(ckpt_path, map_location=device)
    state_dict = checkpoint.get('model_state_dict', checkpoint) if isinstance(checkpoint, dict) else checkpoint.state_dict()
    model.load_state_dict(state_dict, strict=False)
    print('Checkpoint weights loaded successfully!')
else:
    print('Running demonstration mode with initialized architecture.')
```
