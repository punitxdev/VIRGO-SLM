```python
# Clean, vectorized string concatenation
df['pair'] = "<bos>User: " + df['instruction'].astype(str) + "<newline><newline>Assistant: " + df['output'].astype(str) + "<eos>"
```

```python
# Quick check to ensure formatting applied correctly
print(df.sample()['pair'])
```

```python
df['token_length'] = df['pair'].apply(lambda x: len(tokenizer.encode(x)))
```

```python
df_shuffled = df.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
df.sample(15)['token_length']
```

```python
# Flag rows that fit within 1024 tokens
df['under_context_length'] = df['token_length'] <= 1024
```

```python
df.sample(20)[['token_length', 'under_context_length']]
```

```python
df_ideal = df[(df['token_length'] >= 150) & (df['token_length'] <= 500)].reset_index(drop=True)
```

```python
df_ideal.head()
```

```python
import pandas as pd

# 1. Define target columns to keep strictly to instruction and output
target_columns = ['instruction', 'output']
existing_columns = [col for col in target_columns if col in df_ideal.columns]

# 2. Slice the dataframe to keep only what you need
df_export = df_ideal[existing_columns]

# 3. Export the clean dataset to CSV
df_export.to_csv('virgo_coding_IFT.csv', index=False, encoding='utf-8')

print(f"Successfully saved {len(df_export)} rows with columns: {existing_columns} 🚀")
```
